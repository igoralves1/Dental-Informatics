# **Python Collections (Arrays) - List, Tuple, Set, Dictionary**

There are four collection data types in the Python programming language:

- **List** is a collection which is **ordered** and **changeable**. **Allows duplicate members**.  
- **Dictionary** is a collection which is **ordered** and **changeable**. **No duplicate members**.

- **Tuple** is a collection which is **ordered** and **unchangeable**. **Allows duplicate members**.  
- **Set** is a collection which is **unordered**, **unchangeable** elements, and unindexed. **No duplicate members**.  


**Note**
> By **ordered**, it means that each element is assigned an index and can be referenced by that index. An ordered collection **maintains the order of the elements based on the sequence you put stuff into/remove them** from the collection. An ordered collection means that the elements of the collection have a specific order. The **order is independent of the value**. A List is an example.
>
> A **sorted collection** means that not only does the collection have order, but **the order depends on the value of the element**. A SortedSet is an example. A sorted collection **keeps the elements sorted based on a sort criteria**.



1.   **List** : **["elemt1", "elemt2", "elemtN"]** or **[2,56,798]**
> - Lists are used to store multiple items in a single variable.  
> - List items are **ordered**, **changeable**, and **allow duplicate** values. We can change, add, and remove items in a list after it has been created. Since **lists are indexed**, lists can have items with the same value (can be **duplicated**) or with different types.
> - Are ordered collections, meaning that the elements in the list are stored in a specific order. The order of the elements in a list **can be sorted** by using the **sort()** method.   
> - List items are indexed, the first item has index [0], the last item has index [n-1].  
> - Constructor **list()**: It is also possible to use the list() constructor when creating a new list.
> - Lists are created using **[]** square brackets **if not using Constructor list()**:

```python
# Create a list using List Constructor list()
list1 = list(("apple1", "banana1", "cherry1"))
print(list1)

# Create a list without using List COnstructor list(). Using using [] square brackets
thislist = ["apple", "banana", "cherry"]
print(thislist)
print(len(thislist))
print(type(thislist))

thislist2 = list(("apple1", "banana1", "cherry1"))
print(thislist2)
```

2.   **Dictionary**
> - Are are **ordered**, **mutable**, and **Not allow duplicate** values.
> - Dictionary items are presented in key:value pairs, and can be referred to by using the key name.
> - Dictionary items are ordered, changeable, and do not allow duplicates (cannot have two items with the same key:).

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(thisdict)

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(thisdict["brand"])
# print(thisdict[0]) # This will throw an error

# Dictionaries cannot have two items with the same key. Duplicate values will overwrite existing values:
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "year": 2020
}
print(thisdict)
```

3.   **Tuple** : (10, 20, 10) or ("a", "b", "c")
> - Are **ordered**, **allow duplicates**  collections, but they are **immutable**, meaning that the **order of the elements in a tuple cannot be changed**.
> - The elements of a tuple **cannot be changed**, **cannot be added**, or **cannot be removed** after it has been created. This is in contrast to lists, which are mutable, meaning that their elements can be changed.
> - Usage: Tuples are often used to store data that does not need to be changed, such as the coordinates of a point or the names of the days of the week. They can also be used to store data that is passed to a function, as this ensures that the data cannot be accidentally modified by the function.
> - Constructor **tuple()**: It is possible to use the tuple() constructor to make a Tuple.


```python
# Create a tuple using Tuple Constructor tuple()
thistuple = tuple(("apple", "banana", "cherry")) # note the double round-brackets
print(thistuple)

# Create a tuple of coordinates
coordinates = (10, 20, 10)
print(coordinates)
# Try to change the first element of the tuple. This will cause a TypeError, as tuples are immutable
# coordinates[0] = 15

# Create a tuple of days of the week
days_of_the_week = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")
print(days_of_the_week)
# Tuple is ordered meaning is
print(days_of_the_week[0])
print(days_of_the_week[3])

# Try to add a new day to the tuple. This will also cause a TypeError, as tuples are immutable
# days_of_the_week.append("New Day")
```

4.   Set
> - Is a collection of **unordered**, **unique (unindexed)** elements **Not allow duplicates** (duplicated elements are deleted - keep only one). The elements of a set **can be any type of object**, including other sets. **Sets are mutable**, meaning that they can be changed after they are created - can be **added** or have **removed** elements. However, **the elements of a set are immutable**, meaning that they **cannot be changed once they are added** to the set.  
> - If we need to create a set whose elements cannot be added or removed, we need to use the **frozenset()** function. A frozenset is an immutable set. Once a frozenset is created We cannot add or remove elements from a frozenset. We also cannot change the value of an element in a frozenset.
Frozensets are useful for creating sets of data that need to be immutable, such as the set of valid values for a configuration setting.  
> - Sets are unordered, so you cannot be sure in which order the items will appear.  
> - Do not allow duplicate values  
> - Constructor **set()**: It is possible to use the set() constructor to make a set.  
> - Set are created using **{}** square brackets **if not using Constructor list()**:

```python
# Create an empt set and add values
x = set()
x.add(1)
x.add(2)
x.add(2)
print(x)

# Creates a set of numbers:
my_set = {1, 2, 3, 4, 5, 5}
print(my_set)

# Add new elements to the set using the add() method
my_set.add(6)
print(my_set)

# We can also remove elements from the set using the remove() method:
my_set.remove(3)
print(my_set)

# However, we cannot change the value of an element in the set. For example, the following code will not work because the elements of a set are immutable:
# my_set[1] = 10
print(my_set)

# Creates a frozenset of numbers. If we try to add or remove or change an element of a frozenset it will throw an error
my_frozenset = frozenset({1, 2, 3, 4, 5})
my_set.add(6)

# Creates a set using Contructors
thisset = set(("apple", "banana", "cherry")) # note the double round-brackets
print(thisset)

# Creates a set of numbers. This delete duplicates and keep only one for each duplicated element.
my_set2 = {1, 2, 3, 4, 5, 5, 5}
print(my_set2)
```
